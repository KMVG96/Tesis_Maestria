```python


# Cargar shapefile (Quintana Roo)
shapefile_path = "C:/Users/katy/Documents/tesis/GMB_AGBD_Mexico/Codigo/QuintanaRoo_shapefile.shp"
gdf = gpd.read_file(shapefile_path)

# Abrir con el mismo sistema de coordenadas que el raster
with rasterio.open("C:/Users/katy/Documents/tesis/GMB_AGBD_Mexico/Codigo/Forest_height_2019_NAM.tif") as src:
    raster_crs = src.crs

# Reproyectar shapefile al CRS del raster
if gdf.crs != raster_crs:
    gdf = gdf.to_crs(raster_crs)

# generar una mascara para el recorte 
geometries = gdf.geometry.values

# Abrir el raster y aplicar el recorte
raster_path = "C:/Users/katy/Documents/tesis/GMB_AGBD_Mexico/Codigo/Forest_height_2019_NAM.tif"

with rasterio.open(raster_path) as src:
    out_image, out_transform = mask(src, geometries, crop=True)
    out_meta = src.meta.copy()

# Actualizar metadata para el nuevo tamaño del raster recortado
out_meta.update({
    "driver": "GTiff",
    "height": out_image.shape[1],
    "width": out_image.shape[2],
    "transform": out_transform
})

# Guardar el nuevo raster
output_path = "C:/Users/katy/Documents/Tesis_Maestria/Forest_height_QROO.tif"
with rasterio.open(output_path, "w", **out_meta) as dest:
    dest.write(out_image)

print(" Recorte completado y guardado en:", output_path)
```

```python
import matplotlib.pyplot as plt
from rasterio.plot import show

# Abrir el raster recortado
recorte_path = "C:/Users/katy/Documents/Tesis_Maestria/Forest_height_QROO.tif"

with rasterio.open(recorte_path) as src:
    fig, ax = plt.subplots(figsize=(10, 8))
    show(src, ax=ax, cmap='viridis')
    gdf.boundary.plot(ax=ax, color='red', linewidth=1)  # Dibuja el contorno del shapefile
    ax.set_title("Altura Forestal 2019 en Quintana Roo")
    plt.axis('off')
    plt.show()
```

```python
import matplotlib.pyplot as plt
from rasterio.plot import show
import rasterio

# Ruta al raster recortado
recorte_path = "C:/Users/katy/Documents/Tesis_Maestria/Forest_height_QROO.tif"

with rasterio.open(recorte_path) as src:
    data = src.read(1)
    
    # Crear la figura
    fig, ax = plt.subplots(figsize=(10, 8))

    # Mostrar el raster y guardar el objeto de la imagen para la colorbar
    img = ax.imshow(data, cmap='tab20', extent=src.bounds, origin='upper')

    # Añadir shapefile (contorno)
    gdf.boundary.plot(ax=ax, color='red', linewidth=1)

    # Títulos y etiquetas
    ax.set_title("Altura Forestal 2019 en Quintana Roo")
    ax.set_xlabel("Longitud")
    ax.set_ylabel("Latitud")

    # Añadir la barra de color
    cbar = plt.colorbar(img, ax=ax, shrink=0.7)
    cbar.set_label("Altura (m)")

    plt.show()
```
